# Data Overview

## Dataset Description:
 This dataset contains 918 patient records with 12 clinical and demographic attributes used to study factors associated with heart disease.
 Each row corresponds to an individual patient, summarizing demographic attributes, cardiovascular measurements, and diagnostic indicators used to classify heart disease presence (1) or absence (0).

## Column Overview:
Categorical variables: Sex, ChestPainType, RestingECG, FastingBS, ExerciseAngina, ST_Slope


Numerical variables: Age, RestingBP, Cholesterol, MaxHR, Oldpeak   


Target/outcome variable: HeartDisease

## Data Quality Notes:
Missing values? None detected.


Duplicates? No duplicated rows found.


Inconsistent formatting? None observed (categorical labels are clean and standardized).


Outliers? Yes, multiple biologically implausible values detected.

## Immediate Observations:
The dataset is well-structured with no missing values or formatting issues. Heart disease prevalence is roughly balanced (~55% positive). Biological measurements contain impossible zeros, indicating data entry or measurement issues. RestingBP contains 1 value of 0, which is physiologically impossible (blood pressure cannot be 0). Cholesterol contains 172 zeros, also impossible (cholesterol cannot be 0), accounting for ~18% of the dataset. Oldpeak is highly skewed, with over 450 observations recorded as 0.
## Goals for the EDA:
The main objective of this analysis is to understand which patient characteristics are most strongly associated with the presence of heart disease. This includes examining demographic features (such as age and sex), clinical measurements (resting blood pressure, cholesterol, maximum heart rate), and diagnostic test results (ECG findings, exercise-induced angina, ST slope). The goal is not to build a predictive model, but rather to uncover patterns, risk indicators, and meaningful relationships within the data.
## Key Questions to Explore:

1. How does the prevalence of heart disease vary across age groups and between sexes?

2. Which clinical measurements (RestingBP, Cholesterol, MaxHR, Oldpeak) show the strongest differences between heart-disease and non–heart-disease patients?

3. Do specific chest-pain types or ECG patterns correspond to higher risk levels?

4. Does exercise-induced angina and ST-slope behavior meaningfully separate high-risk and low-risk patients?

5. Are there interactions between features (e.g., age × sex, MaxHR × ExerciseAngina) that highlight distinct subgroups of risk?

## Interpretation of Bivariate Findings
The main dataset were used for all categorical bivariate analysis and the dataset excluding cholesterol zero records were used for all numerical analysis.

1. Sex × HeartDisease

Heart disease prevalence is substantially higher among males compared to females. This aligns with known epidemiological trends and suggests sex-specific risk differences within the dataset.

2. Age × HeartDisease

Heart disease becomes more common with increasing age. The distribution shows a noticeable shift toward older age groups among patients with HeartDisease = 1, indicating age is a meaningful risk factor.

3. ChestPainType × HeartDisease

Most HeartDisease cases occur among patients classified with Asymptomatic (ASY) chest pain. In contrast, individuals with Aypical Angina (ATA) show the lowest association with heart disease. This pattern suggests that symptomatic presentation differs significantly between healthy and affected patients.

4. ExerciseAngina × HeartDisease

A strong relationship exists between exercise-induced angina and heart disease. The majority of patients reporting Y (Yes) are in the HeartDisease-positive group, indicating that stress-response symptoms are highly informative.

5. ST_Slope × HeartDisease

The Flat and Down ST slope type is strongly associated with heart disease, while Up slopes dominate among healthy individuals. This makes ST slope one of the clearest discriminators in the dataset.

6. RestingBP × HeartDisease

Resting blood pressure distributions are similar across disease outcomes, with no strong separation between groups. This suggests RestingBP alone may have limited discriminative value after removing the single impossible value (0).

7. Cholesterol × HeartDisease (df1 only)

After excluding impossible cholesterol values, cholesterol levels do not differ meaningfully between outcome groups. No clear pattern emerges, suggesting cholesterol (in this dataset) is not a strong predictor of heart disease.

8. MaxHR × HeartDisease

Maximum heart rate shows an inverse relationship with heart disease—patients with heart disease typically have slightly lower MaxHR values. This aligns with clinical expectations regarding cardiovascular fitness.

9. Oldpeak × HeartDisease

Oldpeak shows strong separation between groups. Higher OP values occur more frequently among patients with heart disease, making Oldpeak one of the most predictive continuous features.

10. RestingECG × HeartDisease

RestingECG patterns differ mildly between disease groups, but the effect is weaker than other diagnostic features. Most observations fall under “Normal” for both outcomes.

```python
# First, check which column contains Interval objects
# Most likely it's one of the numeric columns that was binned
# Let's convert any Interval columns to their midpoint values

num_cols = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak', 'HeartDisease']

# Create a copy of the dataframe with numeric columns
df_numeric = df[num_cols].copy()

# Convert any Interval objects to their midpoint values
for col in num_cols:
    # Check if the column contains any Interval objects
    if pd.api.types.is_interval_dtype(df_numeric[col]) or df_numeric[col].apply(lambda x: isinstance(x, pd.Interval)).any():
        # Extract the midpoint of each interval more safely
        df_numeric[col] = df_numeric[col].apply(
            lambda x: (x.left + x.right)/2 if isinstance(x, pd.Interval) 
            else x if pd.notna(x) else x
        )
    
# Now create the heatmap with the numeric data
sns.heatmap(df_numeric.corr(), annot=True, fmt=".2f", cmap='coolwarm')
plt.show()
```

## Multivariate Synthesis

Bringing the key relationships together reveals a coherent profile of the patients most likely to present with heart disease in this dataset. Several variables that appeared predictive in isolation remain influential when considered jointly.

1. Age, Sex, and Symptom Profile Work Together as Strong Predictors
Older male patients consistently show higher rates of heart disease, and this effect is amplified when paired with high-risk symptom presentations. Males with Asymptomatic or Non-Anginal chest pain types are especially overrepresented among heart-disease cases, suggesting that classical chest pain symptoms are not always present in high-risk individuals.

2. Functional Indicators (ExerciseAngina, MaxHR, ST_Slope, Oldpeak) Form the Strongest Cluster of Predictors
A group of exercise-related diagnostics shows a tightly aligned risk pattern:

ExerciseAngina = Yes,

ST_Slope = Flat or Down,

Lower MaxHR,

Higher Oldpeak,

tend to occur together in HeartDisease-positive patients.
This combination points to reduced cardiovascular efficiency, abnormal stress responses, and ischemic changes—well-established clinical markers that align logically in a multivariate context.

3. Blood Pressure and Cholesterol Are Weak Predictors in This Dataset
After removing impossible values, RestingBP and Cholesterol show minimal separation between disease outcomes. Their contribution is overshadowed by more diagnostic variables (e.g., ECG features, exercise response). This suggests their predictive value is low within this sample, even if clinically relevant in broader populations.

4. Interactions Reveal More Than Any Single Variable Alone
Several patterns only become clear when features are viewed in combination:

Older age + ExerciseAngina = Yes → substantially higher heart-disease risk

ASY chest pain + Flat ST_Slope → strong multivariate signal

Lower MaxHR + Higher Oldpeak → steep increase in heart-disease prevalence

These interactions indicate that risk is not additive but compound. The presence of multiple weak-to-moderate signals often creates a strong overall prediction pattern.

5. Overall Risk Profile From Multivariate Patterns
Patients most associated with heart disease in this dataset tend to share the following combined characteristics:

Male

Older age

Asymptomatic or atypical chest pain presentation

Exercise-induced angina

Flat/Down ST slope

Higher Oldpeak

Lower MaxHR

# Key Questions Answered

## 1. How does the prevalence of heart disease vary across age groups and between sexes?

Heart disease prevalence increases steadily with age, with the highest rates observed in patients aged 50–65+. Additionally, males show a substantially higher prevalence of heart disease than females across nearly all age groups, indicating that age and sex jointly influence risk.

## 2. Which clinical measurements (RestingBP, Cholesterol, MaxHR, Oldpeak) show the strongest differences between heart-disease and non–heart-disease patients?

Among the continuous variables, Oldpeak and MaxHR display the clearest separation between outcome groups. Higher Oldpeak values and lower MaxHR values are strongly associated with heart disease.
RestingBP and cholesterol show minimal discriminative value in this dataset, especially after removing physiologically impossible measurements (0).

## 3. Do specific chest-pain types or ECG patterns correspond to higher risk levels?

Yes. Asymptomatic (ASY) chest pain is strongly associated with heart disease, while ATA (Atypical Angina) is linked to the lowest prevalence.
RestingECG patterns show weaker associations.

## 4. Does exercise-induced angina and ST-slope behavior meaningfully separate high-risk and low-risk patients?

Absolutely. ExerciseAngina (Yes) is strongly concentrated in the heart-disease group, making it a powerful risk indicator.
Similarly, Flat and Down ST slopes correspond to markedly higher disease prevalence, while Up slopes are predominantly found among healthy patients. Together, these exercise-response features provide some of the clearest group separation in the dataset.

## 5. Are there interactions between features (e.g., age × sex, MaxHR × ExerciseAngina) that highlight distinct subgroups of risk?

Yes. Several interactions suggest meaningful subgroups:

Older males show a disproportionately higher rate of heart disease compared to other age–sex combinations.

Low MaxHR combined with ExerciseAngina = Yes identifies a particularly high-risk subgroup.

ASY chest pain paired with Flat/Down ST slope highlights another cluster with elevated disease likelihood.

These interactions suggest that risk is multidimensional and strengthened when evaluating features jointly rather than in isolation.



# Conclusion

This exploratory analysis reveals clear and clinically consistent patterns in the factors associated with heart disease. Demographic variables such as age and sex exhibit strong relationships with the outcome: heart disease is more common among older individuals and is significantly more prevalent in males. Symptom-related features—particularly ChestPainType, ExerciseAngina, and ST_Slope—demonstrate the strongest categorical separation between patients with and without heart disease, marking them as meaningful indicators of cardiovascular stress and risk.

Among numerical variables, Oldpeak and MaxHR show the most substantial differences between outcome groups. Higher Oldpeak values and lower maximum heart rates are more common among heart-disease patients. After removal of impossible zero values, Cholesterol shows minimal distinction between groups, while RestingBP exhibits no meaningful separation.

Multivariate patterns reinforce the individual findings: combinations such as older males, patients with exercise-induced angina, and those displaying flat/down ST slopes consistently form higher-risk subgroups. These interactions suggest that heart-disease likelihood increases not through a single feature, but through the convergence of age, symptoms, and cardiac-stress indicators.

Overall, the dataset presents coherent and interpretable relationships. The insights gained offer a clear foundation for future predictive modeling or targeted clinical analysis.
